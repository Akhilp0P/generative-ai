# ESG Keyword Extraction — EDA

Explore ESG disclosure structure, document volume, vocabulary and ESG category coverage.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('data/esg_documents.csv')
df.head()
```

```python
df.info()
print('Companies:', df.company.nunique())
print('Years:', df.year.nunique())
print('Document types:', df.document_type.value_counts())
```

```python
df['word_count'] = df.text.str.split().str.len()
df[['company','document_type','section','word_count']].sort_values('word_count', ascending=False)
```

```python
top_terms = (df.text.str.lower().str.replace(r'[^a-z0-9\s-]', ' ', regex=True).str.split().explode().value_counts().head(20))
top_terms.plot(kind='bar', title='Most Frequent Terms in Sample ESG Corpus')
plt.tight_layout()
```
